```python
# -*- coding: utf-8 -*-
"""
Análisis de series temporales de fast fashion:
- Limpieza y preparación de datos
- Análisis descriptivo
- Visualización de series
- Análisis de estacionalidad (descomposición)
"""

import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose

# -------------------------------------------------------------------
# 1. Carga de datos
# -------------------------------------------------------------------
# Cambia la ruta por la ubicación de tu fichero
file_path = "fastfashion.xlsx"

df = pd.read_excel(file_path)

# -------------------------------------------------------------------
# 2. Limpieza y preparación
# -------------------------------------------------------------------
# Convertir la columna de mes a formato fecha (año-mes)
df["Mes"] = pd.to_datetime(df["Mes"], format="%Y-%m")

# Asegurar que las columnas numéricas están en formato adecuado
df["zara"] = pd.to_numeric(df["zara"], errors="coerce")
df["temu"] = pd.to_numeric(df["temu"], errors="coerce")
df["shein"] = pd.to_numeric(df["shein"], errors="coerce")

# Definir Mes como índice temporal
df = df.set_index("Mes").sort_index()

# -------------------------------------------------------------------
# 3. Análisis descriptivo básico
# -------------------------------------------------------------------
print("\n=== Descriptivos básicos ===")
print(df[["zara", "temu", "shein"]].describe())

# -------------------------------------------------------------------
# 4. Gráfico de la evolución temporal
# -------------------------------------------------------------------
plt.figure(figsize=(12, 6))
plt.plot(df.index, df["zara"], label="Zara")
plt.plot(df.index, df["shein"], label="Shein")
plt.plot(df.index, df["temu"], label="Temu")

plt.title("Evolución temporal de fast fashion")
plt.xlabel("Fecha")
plt.ylabel("Valor")
plt.legend()
plt.tight_layout()
plt.show()

# -------------------------------------------------------------------
# 5. Descomposición estacional (modelo aditivo, frecuencia mensual)
# -------------------------------------------------------------------
# NOTA: period=12 asume estacionalidad anual con datos mensuales

results = {}
for col in ["zara", "shein", "temu"]:
    serie = df[col].dropna()

    # Nos aseguramos de que haya datos suficientes
    if len(serie) < 24:
        print(f"\n[AVISO] Serie demasiado corta para descomponer: {col}")
        continue

    print(f"\n=== Descomposición estacional para {col} ===")

    res = seasonal_decompose(serie, model="additive", period=12)
    results[col] = res

    # Graficar solo la componente estacional
    plt.figure(figsize=(12, 4))
    plt.plot(res.seasonal)
    plt.title(f"Componente estacional mensual - {col}")
    plt.xlabel("Fecha")
    plt.ylabel("Componente estacional")
    plt.tight_layout()
    plt.show()

    # Si quieres ver las cuatro partes completa (observed, trend, seasonal, resid):
    # res.plot()
    # plt.suptitle(f"Descomposición completa - {col}")
    # plt.tight_layout()
    # plt.show()

# -------------------------------------------------------------------
# 6. Estacionalidad por mes (media mensual de todo el período)
# -------------------------------------------------------------------
# Creamos una columna con el número de mes (1-12)
df_monthly = df.copy()
df_monthly["mes_num"] = df_monthly.index.month

media_mensual = (
    df_monthly.groupby("mes_num")[["zara", "shein", "temu"]]
    .mean()
    .sort_index()
)

print("\n=== Media mensual por marca (estacionalidad promedio) ===")
print(media_mensual)

# Gráfico de la media mensual para cada marca
plt.figure(figsize=(10, 5))
plt.plot(media_mensual.index, media_mensual["zara"], marker="o", label="Zara")
plt.plot(media_mensual.index, media_mensual["shein"], marker="o", label="Shein")
plt.plot(media_mensual.index, media_mensual["temu"], marker="o", label="Temu")

plt.xticks(range(1, 13))
plt.title("Estacionalidad media mensual (promedio de todos los años)")
plt.xlabel("Mes")
plt.ylabel("Media del indicador")
plt.legend()
plt.tight_layout()
plt.show()
```
